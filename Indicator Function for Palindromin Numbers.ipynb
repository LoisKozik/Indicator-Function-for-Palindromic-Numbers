# Indicator Function for Palindromic Numbers

```python
from math import *
import matplotlib.pyplot as plt

epsilon = 1e-8
```

### Mathematical Solution

```python
def N(n):
    return floor(log10(n) + 1)

def f(a,b):
    return floor((a + epsilon)/(b + epsilon)) * floor((b + epsilon)/(a + epsilon))

def C(n,d):
    return floor(10*(n/10**d - floor(n/10**d) + epsilon))

def D_1(n):
    S = 0
    for k in range(1, N(n)+1):
        S += f(C(n,k), C(n,N(n)+1-k))
    return floor(1/N(n) * S)

def D_2(n):
    P = 1
    for k in range(1, N(n)+ 1):
        P = P * f(C(n,k), C(n,N(n)+1-k))
    return P
```

### Numerical verification

```python
def exact(n):
    s = str(n)
    for i in range(len(s)):
        if s[i] != s[len(s)-1-i]:
            return 0
    return 1
```

```python
n1 = 1456
n2 = 143341

print(D_1(n1), D_1(n2))
print(D_2(n1), D_2(n2))
```

### Final Results

```python
n = int(1e6)
L = L = [D_1(i) == exact(i) for i in range(1,n+1)]

print("Percentage of success", sum(L)/n * 100)
```
